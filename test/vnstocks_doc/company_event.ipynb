```python
from vnstock import Listing

def get_ticker_list():
    try:
        df = Listing().all_symbols()
        tickers = df["symbol"].astype(str).str.upper().str.strip().unique().tolist()
    except Exception as e:
        print(f"Lỗi lấy danh sách mã: {e}")
        tickers = []

    return tickers
```

```python
list_cp = get_ticker_list()
print(f"Số lượng mã: {len(list_cp)}")
```

```python
from vnstock import Company
import pandas as pd
import time

def get_company_events(symbols, delay=0.5, batch_delay=2):
    all_data = []
    
    for idx, symbol in enumerate(symbols):
        try:
            company = Company(symbol=symbol, source='VCI')
            data_list = company.events()
            
            # Kiểm tra xem data_list có dữ liệu không
            if data_list is not None and not data_list.empty if isinstance(data_list, pd.DataFrame) else bool(data_list):
                if isinstance(data_list, pd.DataFrame):
                    for _, item in data_list.iterrows():
                        record = {
                            'event_title': item.get('event_title') if hasattr(item, 'get') else item['event_title'],
                            'public_date': item.get('public_date') if hasattr(item, 'get') else item['public_date'],
                            'source_url': item.get('source_url') if hasattr(item, 'get') else item['source_url'],
                            'event_list_name': item.get('event_list_name') if hasattr(item, 'get') else item['event_list_name'],
                            'event_list_code': item.get('event_list_code') if hasattr(item, 'get') else item['event_list_code'],
                            'id': item.get('id') if hasattr(item, 'get') else item['id']
                        }
                        all_data.append(record)
                else:
                    for item in data_list:
                        record = {
                            'event_title': item.get('event_title'),
                            'public_date': item.get('public_date'),
                            'source_url': item.get('source_url'),
                            'event_list_name': item.get('event_list_name'),
                            'event_list_code': item.get('event_list_code'),
                            'id': item.get('id')
                        }
                        all_data.append(record)
            print(f"✓ {symbol} ({idx+1}/{len(symbols)})")
        except Exception as e:
            print(f"✗ {symbol}: {e}")
        
        if idx < len(symbols) - 1:
            time.sleep(delay)
        
        if (idx + 1) % 19 == 0:
            print(f"⏸ Nghỉ {batch_delay}s sau {idx + 1} mã...")
            time.sleep(batch_delay)
    
    if all_data:
        df = pd.DataFrame(all_data)
        return df
    else:
        return pd.DataFrame()

```

```python
df_events = get_company_events(list_cp, delay=5.5, batch_delay=5)
print(f"\nTổng số sự kiện: {len(df_events)}")
print(df_events.head())
```
